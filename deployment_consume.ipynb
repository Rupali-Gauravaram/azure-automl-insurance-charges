```python
import json
import os
import urllib.request

# Fixed input schema (independent variables)
columns = [
    "age",
    "sex",
    "bmi",
    "children",
    "smoker",
    "Claim_Amount",
    "past_consultations",
    "num_of_steps",
    "Hospital_expenditure",
    "NUmber_of_past_hospitalizations",
    "Anual_Salary",
    "region"
]

# Example: Replace this with your actual row of values
# Order must match the columns above
values = [27, "male", 27, 2, "yes", 16000, 5, 8000, 45000, 1, 140000, "northwest"]

data = {
    "input_data": {
        "columns": columns,
        "data": [values]   # Can pass multiple rows like [[row1], [row2], ...]
    }
}

body = json.dumps(data).encode("utf-8")

# Your Azure endpoint(url)
url = os.environ["AZURE_ML_ENDPOINT"]  # e.g. https://<endpoint>.<region>.inference.ml.azure.com/score

# Your API key(primary key)
api_key = os.environ["AZURE_ML_KEY"]  # never hard-code the key

# Headers for authentication
headers = {
    "Content-Type": "application/json",
    "Authorization": f"Bearer {api_key}"
}

# Request
req = urllib.request.Request(url, body, headers)

try:
    response = urllib.request.urlopen(req)
    result = response.read()
    print("Prediction Charges Result: ")
    print(result.decode("utf-8"))
except urllib.error.HTTPError as error:
    print("Error: ")
    print(error.info())
```
